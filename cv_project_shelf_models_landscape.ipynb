## # 🛒 Retail Shelf Intelligence System using YOLOv8

## 📌 Introduction

Computer Vision has become one of the fastest-growing fields of Artificial Intelligence, enabling machines to understand and interpret visual information from images and videos. In the retail industry, computer vision is increasingly used for inventory monitoring, shelf auditing, product recognition, and store analytics to improve operational efficiency and reduce manual inspection efforts.

This project presents an end-to-end **Retail Shelf Intelligence System** that automatically detects and analyzes objects present on supermarket shelves using the YOLOv8 object detection framework. The model is trained through transfer learning on a custom retail shelf dataset containing four object categories: **Product, Price Card, Shelf, and POP (Point of Purchase Display)**. In addition to object detection, the project performs retail analytics by counting detected objects, generating business insights, tracking products across video frames using ByteTrack, and exporting structured reports for further analysis.

The complete project is developed in Kaggle using NVIDIA Tesla T4 GPUs and demonstrates the complete computer vision workflow from data analysis and model training to inference, tracking, and business intelligence generation.

---

# 📖 Project Summary

This project develops a complete AI-powered retail shelf monitoring solution capable of automatically identifying products and related retail elements from both images and videos.

The workflow begins with dataset exploration and preprocessing, followed by training multiple YOLOv8 models using transfer learning. After comparing the performance of different model variants, the best-performing model is selected and evaluated using standard object detection metrics including Precision, Recall, F1-Score, mAP@50, and mAP@50-95.

The trained model is then deployed for image and video inference, where detected objects are counted, analyzed, and converted into meaningful retail statistics such as product count, price card count, shelf utilization, and promotional display detection. Finally, ByteTrack is integrated for video object tracking, enabling unique object identification across frames and reducing duplicate counting.

The result is a practical Computer Vision application that combines deep learning with business analytics for intelligent retail monitoring.

# #Extract Detection Information

````python
import pandas as pd

detections = []

for i, box in enumerate(results[0].boxes):

    cls = int(box.cls[0])
    conf = float(box.conf[0])

    x1, y1, x2, y2 = box.xyxy[0].tolist()

    width = x2 - x1
    height = y2 - y1

    center_x = (x1 + x2) / 2
    center_y = (y1 + y2) / 2

    area = width * height

    detections.append({
        "Object ID": i + 1,
        "Class": model.names[cls],
        "Confidence": round(conf, 3),
        "X1": round(x1),
        "Y1": round(y1),
        "X2": round(x2),
        "Y2": round(y2),
        "Width": round(width),
        "Height": round(height),
        "Area": round(area),
        "Center X": round(center_x),
        "Center Y": round(center_y)
    })

df = pd.DataFrame(detections)

df.head(15)
````

## Extract Every Detection

````python
import pandas as pd

tracking_data = []

for frame_result in results:

    frame_number = frame_result.path

    if frame_result.boxes is None:
        continue

    for box in frame_result.boxes:

        cls = int(box.cls[0])

        conf = float(box.conf[0])

        track_id = -1

        if box.id is not None:
            track_id = int(box.id[0])

        x1,y1,x2,y2 = box.xyxy[0].tolist()

        tracking_data.append({

            "Track ID":track_id,

            "Class":model.names[cls],

            "Confidence":round(conf,3),

            "x1":round(x1),

            "y1":round(y1),

            "x2":round(x2),

            "y2":round(y2)

        })

tracking_df = pd.DataFrame(tracking_data)

tracking_df.head()
````

## REFLECTION AND CONCLUSION

1. Why did you choose this problem?

I chose retail shelf monitoring because it is a practical real-world computer vision application. Retail stores need efficient methods to monitor products, price cards, and promotional displays. Automating this process can reduce manual effort, improve inventory visibility, and support better retail decision-making.

2. What was the most difficult part of data collection or annotation?

The biggest challenge was working with an imbalanced dataset. The Product and Price Card classes had tens of thousands of instances, while the POP class had only a few hundred examples. This imbalance made it difficult for the model to learn minority classes effectively and affected overall performance.

3. Why did you choose your specific YOLO model?

I initially trained YOLOv8s as a lightweight baseline because it trains quickly and establishes reference performance. After evaluating the results, I selected YOLOv8m with an image size of 800×800 because it provided better feature extraction and improved recall and mAP, making it more suitable for detecting small and densely packed retail objects.

4. Which class performed best and which performed worst?

The Shelf class achieved the best performance with an mAP@50 of approximately 0.698, indicating strong detection capability.

The Price Card class performed the worst, with an mAP@50 of approximately 0.173, mainly due to its small size, visual similarity across instances, and the complexity of crowded shelf scenes. The POP class also remained challenging because of the limited number of training examples.

5. What types of errors did your model make?

The model occasionally missed small objects such as Price Cards and POP displays. It also produced lower-confidence detections for partially occluded products and sometimes failed to detect shelves in images that differed significantly from the training distribution. These errors were primarily caused by class imbalance, small object size, and domain differences.

6. How did your model perform on unseen images or videos?

The model generalized reasonably well to unseen supermarket images and videos. Product detection remained consistent, while Price Card detection improved when using a lower confidence threshold during visualization. Video tracking successfully maintained object identities using ByteTrack, enabling unique object counting and reducing duplicate detections across frames.

7. What would you improve with more time, data, or compute?

With additional resources, I would:

Collect a larger and more diverse retail dataset.
Increase the number of POP and Shelf annotations.
Experiment with newer YOLO versions or larger models.
Explore instance segmentation for more accurate shelf analysis.
Optimize hyperparameters using automated search techniques.
Deploy the solution as a real-time web application using Streamlit or Gradio.
8. How could this system be used in a real-world setting?

The system could be integrated into retail stores using CCTV cameras or mobile inspection devices. It could automatically monitor shelf availability, verify price card placement, detect promotional displays, assist inventory audits, and generate analytics for retail managers. Such a solution could improve operational efficiency, reduce manual inspections, and support data-driven retail management.
